```python
from knowledge_ratchet import run_s4ai, run_periodic, run_nolearn, run_random, run_oracle_threshold
import matplotlib.pyplot as plt

R = 10_000
s4  = run_s4ai(R=R)
per = run_periodic(R=R)
nl  = run_nolearn(R=R)
rnd = run_random(R=R)
orc = run_oracle_threshold(R=R)
print(f"S4AI K_100 = {s4['K_final']:.0f} | oracle ceiling {orc['K_final']:.0f} | "
      f"periodic {per['K_final']:.0f} | no-learn {nl['K_final']:.0f} | random {rnd['K_final']:.0f}")
```

```python
plt.figure(figsize=(6,3.5))
plt.plot(s4['K_mean'],  label='S4AI (RLPE)', lw=2)
plt.plot(orc['K_mean'], label='Oracle threshold ceiling', ls=':')
plt.plot(per['K_mean'], label='Periodic retraining')
plt.plot(nl['K_mean'],  label='No learning')
plt.plot(rnd['K_mean'], label='Random', ls='--')
plt.xlabel('round t'); plt.ylabel('cumulative discoveries K_t')
plt.legend(frameon=False); plt.title('The Knowledge Ratchet'); plt.show()
```
